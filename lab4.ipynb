```python
#read and write to text files using python
students=[]
with open('grades.txt','r') as file:
    for line in file:
        line = line.strip() #removes white space
        if not line:        #if string is empty
            continue        #continues to next line
        name,grade=line.split(",")   #divides a string where comma appears
        student={"Name":name,"Grade":int(grade)}
        students.append(student)

print(students)
        
```

```python
#calculate and print the average grade
grades=[s['Grade'] for s in students] #list comprehension
average_grade=sum(grades)/len(grades)
print(average_grade)
```

```python
# save  a new file `above_average.csv` containing only the names and grades of students
#who scored above average.
if students:  #checks if student list contains items
    with open('above_average.csv','w') as f:
        f.write('Name,Grade\n') #writes csv header row
        for student in students:
            if student["Grade"]>average_grade:
                f.write(f"{student['Name']},{student['Grade']}\n")
                
    print("file 'above_average.csv' has been generated successful")
else:
    print("file 'grades.txt' was empty")



        
    
```

```python
#Wrapping read and write text files 
students=[]
try:    #wraps file opening process
    with open('grades.txt','r') as file:
        for line in file:
            line=line.strip()
            if not line:
                continue
            try:
                name,grade=line.split(",")
                student={"Name":name,"Grade":int(grade)} #structures the data into dictionary
                students.append(student)
            except ValueError:
                print("value error")
except FileNotFoundError:
    print(f"error:file not found")

print(students)
    

            
```

```python
#wrapping calculate and print average grade
if students:
    grades = [s['Grade'] for s in students]
    average = sum(grades) / len(grades)
    print(average)
else:
    print("No valid student records found — cannot calculate an average.")
    average = 0   #sets a default fallback value
```

```python
#wrapping save a new file
try:
    if students:
        with open('above_average.csv', 'w') as file:
            file.write('Name,Grade\n')
            for student in students:
                if student['Grade'] > average_grade:
                    file.write(f"{student['Name']},{student['Grade']}\n")
        print("File 'above_average.csv' has been generated successfully.")
    else:
        print("The file grades.txt was empty.")
except OSError as e:
    print(f"Error writing above_average.csv: {e}")
```

```python
#wrap saving data into json file format
import json

try:
    with open("students.json", "w") as file:
        json.dump(students, file, indent=4)
    print("File 'students.json' has been generated successfully.")
except OSError as e:
    print(f"Error writing students.json: {e}")
```
