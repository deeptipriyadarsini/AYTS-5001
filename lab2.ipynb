```python
Task 3: Password Strength Checker
```

```python
def check_password(password):
    special_characters="!@#$%^&*"
    has_len=len(password)>=8
    has_digit=any(char.isdigit() for char in password)
    has_special=any(char in special_characters for char in password)

    if has_len and has_digit and has_special:
        print("password is strong")
    else:
        if not has_len:
            print("enter minimum eight")
        elif not has_digit:
            print("enter atleast one digit")
        elif not has_special:
            print("atleat one special character")

password=(input("enter password"))
check_password(password)
    
```

```python
Task 5: Temperature Conversion Table 
```

```python
def celsius_to_fahrenheit_table(start,end):
    for celsius in range(start,end+1):
        fahrenheit=(celsius*9/5)+32
        print(f"celsius:{celsius}\tFahrenheit={fahrenheit}")

start_temp=int(input("enter start temperature in celsius"))
end_temp=int(input("enter end temperature in celsius"))
celsius_to_fahrenheit_table(start_temp,end_temp)
```

```python
ask 7: Prime Number Checker
```

```python
def is_prime(n):
    if n<2:
        return False
    for i in range(2,int(n**0.5)+1):
        if n%i==0:
            return False
    return True

number=int(input("enter a number"))
is_prime(number)
        
```
