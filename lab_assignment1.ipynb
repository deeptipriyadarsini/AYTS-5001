```python
CREATING A LIST OF DICTIONARIES
```

```python
#Creating a list with 3 student dictionaries.
students=[] #empty list to hold the dictionaries.
for i in range(1,4):
    print("Enter student details:")
    Name=input("Enter student name:")
    Age=int(input("Enter student age:"))
    Grade=int(input("Enter student grade:"))
    #creating a dictionary.
    student={"Name":Name,"Age":Age,"Grade":Grade}
    #Adding the dictionaries to the list.
    students.append(student)

#printing title.
print(f"Student List:")
#printing list of dictionaries.
print(students)
    
    
            
    
    
```

```python
FINDING AVERAGE OF GRADES
```

```python
#printing average grade.
total_grade=sum(student['Grade'] for student in students)#Generator Expression: Uses parentheses () and generates items one at a time without building a full list in memory.
average_grade=total_grade//len(students)
print(average_grade)
```

```python
NAMES OF STUDENTS WITH GRADES ABOVE AVERAGE
```

```python
#print the name of students who scored above the average
for student in students:
    if student['Grade']>74:
        print(f"{student['Name']} got above average")
```

```python
ALL NAMES IN UPPERCASE
```

```python
#convert all names into upper case and print them.
for student in students:
    student['Name']=student['Name'].upper()
    print(student['Name']) 
```

```python
ERROR HANDLING
```

```python
#ask the user for student name and use try/except to check.
search_name=input("Enter a name:")
try:
    found_name=None 
    for student in students:
        if student['Name'].upper()==search_name.upper():
            found_name=student
            break
    #if not found we trigger the exception.
    if found_name is None:
        raise ValueError(f"student '{search_name}' does not exist in the list")
    if found_name is student:
        print(f"student {search_name}'s grade is {found_name['Grade']}")

except ValueError as error:
    print(f"Error:error")
    


    
```
