```python
#1) Read the `grades.txt` file and parse each line into a dictionary with keys: name and grade.
students = []
try:
    with open('grades.txt','r') as file:
        for line in file:
            line = line.strip() #removes whit space
            if not line:  #skips a line if blank
                continue
            name,grade = line.split(",")
            student = {'Name':name,'Grade':grade}
            students.append(student)
            print(students)

except FileNotFoundError:
  print("Error: The file 'grades.txt' was not found.")
  students = [] #fall back so program doesnt crash


            
        
```

```python
#2. Use the `https://api.agify.io?name=FIRSTNAME` API to estimate the age of each student.
# Extract only the first name for the API query
#Add the predicted age as a new field
import requests
import os
from dotenv import load_dotenv

load_dotenv()
api_key = os.environ["apikey"]


def get_predicted_age(name):
    first_name=name.split()[0]  #slicing from first position
    url="https://api.agify.io"
    params = {"name": first_name, "apikey": api_key}
    response=requests.get(url,params=params)
    return response.json().get('age')

for student in students:
    print(f"student name={student['Name']}, age: {get_predicted_age(student['Name'])}") #concantation,student['Name'] gets the name 
    

for student in students:
    print(student)

    
    


    
    
    
```
